```python
import polars as pl
import numpy as np

# Create a tiny test DataFrame
df = pl.DataFrame({
    "cat": pl.Series(["A", "B", "C", "D"]).cast(pl.Categorical)
})
print("Original DataFrame:")
print(df)
print("\nSchema:", df.schema)

# Try to modify one value
modified = df.clone()
print("\nTrying modification...")
print(modified["cat"].cast(pl.Utf8))
```
